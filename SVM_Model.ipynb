```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
import seaborn as sns

from sklearn.model_selection import train_test_split
```

```python
# Tips Dataset from seaborn library
df= sns.load_dataset('tips')
df
```

```python
# Storing Features and Target Cloumn
x = df.iloc[:,1:]
y = df.iloc[:,0]
```

# Splitting Training And Testing Dataset

```python

X_train , X_test ,y_train , y_test = train_test_split(x,y,test_size = 0.2, random_state = 42 )
```

```python
X_train
```

# Prepocess The  Features Columns (Sex , Smoker , Time)

```python

from sklearn.preprocessing import LabelEncoder
lr1 = LabelEncoder()
lr2 = LabelEncoder()
lr3 = LabelEncoder()

X_train['sex']= lr1.fit_transform(X_train['sex'])
X_train['smoker']=lr2.fit_transform(X_train['smoker'])
X_train['time']=lr3.fit_transform(X_train['time'])
```

```python
X_test['sex']= lr1.transform(X_test['sex'])
X_test['smoker']=lr2.transform(X_test['smoker'])
X_test['time']=lr3.transform(X_test['time'])
```

# One-Hot-Encodeing OF The Feature Column (Day) Using ColumnTransforemer

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
CT =ColumnTransformer(transformers= (
    [('onehot', OneHotEncoder(drop = 'first'),[3])] ), remainder = 'passthrough')
```

```python
X_train = CT.fit_transform(X_train)
```

```python
X_test= CT.transform(X_test)
```

# SVR Model selection and Train The Model

```python
from sklearn.svm import SVR
```

```python
svr = SVR(kernel='rbf')
svr.fit(X_train , y_train)
```

# Predict The Model

```python
predict = svr.predict(X_test)
```

# For The Evaluation Model Using r2_score , mean_absulate_error

```python
from sklearn.metrics import r2_score , mean_absolute_error
```

```python
print(f'r2_score {r2_score(y_test , predict)}')
print(f'mean_square_error: {mean_absolute_error(y_test , predict)}')
```

# For Better Accuracy Using GridSearchCV

```python
from sklearn.model_selection import GridSearchCV as gd
```

```python
params = {'C': [0.1, 1, 10, 100, 1000],
              'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
              'kernel': ['rbf']}
```

```python
gdc = gd(estimator=SVR(), param_grid=params , n_jobs=-1)
```

```python
gdc.fit(X_train ,y_train)
```

```python
pred = gdc.predict(X_test)
```

```python
print(f'r2_score at gridsearch_cv : {r2_score(y_test , pred)}')
print(f'mean_square_error at gridsearch_cv : {mean_absolute_error(y_test , pred)}')
```

```python
print(f'Best parameters: {gdc.best_params_}')
```
